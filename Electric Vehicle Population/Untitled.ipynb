### Task 7 — Handle numerical missing values and Task 8 — Handle categorical missing values

```python
df["2020 Census Tract"] = df["2020 Census Tract"].fillna(value=df["2020 Census Tract"].mode()[0])
df["Electric Range"] = df["Electric Range"].fillna(value=df["Electric Range"].median())
```

```python
df["County"] = df["County"].fillna(value=df["County"].mode()[0])
df["City"] = df["City"].fillna(value=df["City"].mode()[0])
```

```python
df["Postal Code"] = df["Postal Code"].fillna(value=df["Postal Code"].mode()[0])
df["Vehicle Location"] = df["Vehicle Location"].fillna(value=df["Vehicle Location"].mode()[0])
```

```python
df["Electric Utility"] = df["Electric Utility"].fillna(value=df["Electric Utility"].mode()[0])
```

```python
df["Legislative District"] = df["Legislative District"].fillna(value=df["Legislative District"].mode()[0])
```

### Task 16 — Detect outliers using IQR

```python
Q1 = df['Electric Range'].quantile(0.25)
Q3 = df['Electric Range'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
print("Lower Bound:", lower)
print("Upper Bound:", upper)
df_iqr = df[
(df['Electric Range'] >= lower) &
(df['Electric Range'] <= upper)
]
print(df_iqr.head())
```

```python
Q1 = df['2020 Census Tract'].quantile(0.25)
Q3 = df['2020 Census Tract'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
print("Lower Bound:", lower)
print("Upper Bound:", upper)
df_iqr = df[
(df['2020 Census Tract'] >= lower) &
(df['2020 Census Tract'] <= upper)
]
print(df_iqr.head())
```

### Task 21 — Apply outlier capping

```python
import numpy as np
```

```python
Q1 = df['Electric Range'].quantile(0.25)
Q3 = df['Electric Range'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df['Electric Range'] = np.where(
    df['Electric Range'] > upper,
    upper,
    np.where(
        df['Electric Range'] < lower,
        lower,
        df['Electric Range']
    )
)
```
